# Timeseries forecasting / Text classification

### Pick **one** of these two topics.

---

## 1. Time Series with RNNs

1. Download and prepare the timeseries data as seen in the "Download the data" [here](https://github.com/jchwenger/DLWP/blob/main/lectures/07/chapter13_timeseries-forecasting.ipynb).
2. Build an LSTM network, importing code from the first recurrent model in "Recurrent neural networks". 
3. Experiment with different hyperparameters to improve your results, as seen in "Advanced use of recurrent neural networks".
4. Suggestions for experiments:
    - Adjust the number of units in each recurrent layer;
    - Try different learning rates;
    - Substitute LSTM layers for GRU layers;
    - Try adding a recurrent regularizer (you can try recurrent dropout, but it's really slow);
    - Try bigger densely connected regressors on top of the recurrent base, or even a stack of dense layers;
    - Try make your model bidirectional;    
    - Try and reproduce the non-recurrent results from "A common-sense, non-machine-learning baseline", "Training and evaluating a densely connected model" and "Let's try a 1D convolutional model" and compare those with what you have:
      - The common sense baseline;
      - The feedforward net;
      - The 1D convnet.

## 2. Sets vs Sequences

Here again there are various experiments available (no need to try to do *everything*: pick only the line of enquiry you find the most interesting).

1. Download and prepare the IMDB data as seen in "Preparing the IMDB movie reviews data" [here](https://github.com/jchwenger/DLWP/blob/main/lectures/07/chapter14_text-classification.ipynb)
2. N-grams: compare the performance of various set models as seen in "Set models" (unigram, bigram, trigram, etc.). Apart from the usual hyperparameters, key elements to look at are:
    - The vocabulary size (`max_tokens`);
    - The size of the ngrams;
    - The standardization/splitting step.
3. Word embeddings:
    - Train and compare multiple CBOW models as seen in the lecture. The main hyperparameter here is of course the `hidden_dim`.
    - The resulting embeddings can be plugged into other models. One can then measure the impact on performance.
    - Another line of enquiry is the 'manual' exploration of the latent space, using similarity and analogies. Here, it would be interesting to have a few pre-trained embedding matrices (with different `hidden_dim`s), and see if the vectors trained on IMDB display similar properties as the GLOVE ones.
