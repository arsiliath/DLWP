## 1. Interpreting what ConvNets learn

There are three possible pathways for experiments:

1. Visualizing intermediate activations.
  - Make sure you are able to run the code to display the activations in the various layers.
  - This pipeline does not require an enormous model to work, it is possible to train a small ConvNet on another dataset (e.g. FashionMNIST) and see how one of those images is processed.
  - One interesting experiment would be to train a new convent on cats & dogs but replacing the ReLU with a more [recent activation function](https://keras.io/api/layers/activations/), such as `elu`, `gelu`, `silu` (aka `swish`) or `mish`. Those should not produce that many dead units.

3. Visualizing ConvNet filters.
  - In the lecture, only one layer is displayed, and the entire pipeline has to be re-run to change that. It might be interesting to write a function that takes the layer as an argument, runs the pipeline, and displays the result.
  - State-of-the-art models can be inspected in [OpenAI's Microscope](https://openai.com/index/microscope/).
  - Advanced: something quite tricky, but that could be interesting to do, would be to try and learn filter patterns not just for one layer, but for the entire net, in one go. That would require changing the loss function as well, so that the appropriate gradient updates are performed!
  

4. Visualizing heatmaps of class activation.
  - The first line of experiment here is to port the pipeline, then try it on different images. It would be interesting to automate the code so that it is possible to place a number of images in a folder, then run a function that loads each image in turn, applies the pipeline, and saves the result in another folder.
  - Another interesting experiment would be to compare various models from [the ones available out of the box in Keras](https://keras.io/api/applications/).

## 2. Image segmentation

1. Training a model from scratch:
    - Make sure you understand how the IoU metric. Import the code, and the test sections, and experiment with those.
    - Make sure you understand the baseline. Experiment with various sizes of dummy boxes.
    - Experiment with various hyperparameters: the main task here would be to reduce overfitting.
    - Can you encapsulate your ConvNet model so that it is easier to change its architecture (number of layers, kernel size, etc.)?
3. Test the pre-trained model pipeline as [described here](https://deeplearningwithpython.io/chapters/chapter11_image-segmentation/#using-a-pretrained-segmentation-model).

Alleyways for research (e.g. CW):
    - One would be how to create a data augmentation pipeline that applies also to the segmentation targets.

## 3. Object detection

**Beware**: the COCO dataset is fairly large.

1. Training a model from scratch:
    - Make sure you understand how the combined loss works, as well as IoU. Import the code from those, and the test sections, and experiment with those.
    - Make sure you understand the baseline. Experiment with various sizes of dummy boxes.
    - Experiment with various hyperparameters.
        - What happens if you train for longer than 4 epochs?
        - Given the size of the dataset, it might be best to make experiments on the smaller selection, and only scale up with a good set of hyperparameters (found on the small set).
    - Can you encapsulate your ConvNet model so that it is easier to change its architecture (number of layers, kernel size, etc.)?
3. Testing the pre-trained model pipeline as [described here](https://deeplearningwithpython.io/chapters/chapter12_object-detection/#using-a-pretrained-retinanet-detector).

Alleyways for research (e.g. CW):
    - Given a trained model, look into techniques to implement non-max suppression (to get e.g. the top prediction per class, and beyond that the highest-confidence box predictions).
    - Another one would be how to create a data augmentation pipeline that applies also to the grid cells and boxes.
