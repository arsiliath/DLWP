# Language models and the Transformer

## Sequence-to-sequence learning with a Transformer

First off, make sure you run the `np.einsum` examples from Chapter 15 and understand its syntax.

Then port the code to set-up the dataset.

There are three lines of experiments you can take here:

1. High-level experiments, as performed in the past:
    - Using `keras_hub` layers rather than the manual implementation, experiment with various hyperparameters: number of layers, number of heads, embedding dimension, intermediate dimension, use of dropout. I recommend training very small models so that the iteration speed is faster.
    - Can you encapsulate your Transformer model into a function or class so that it is easier to change its architecture (number of layers, number of heads, etc.)?
    - Train an RNN variant (LSTM or GRU) on the same dataset, and compare its performance with your Transformer. It would be interesting to show that smaller variants of both RNNs and Transformers have the same performance relationship (remember that our Transformer with `TokenAndPositionEmbedding` had less than half the amount of parameters as the GRU, and performed better!).
    - There are a few other types of [encoding layers](https://keras.io/keras_hub/api/modeling_layers/) that could be tested to replace `TokenAndPositionEmbedding`.
2. Train on multiple languages:
   - Also using only the `keras_hub` layers (and perhaps still encapsulating the model creation into a function/class, for easier tweaking), you can focus on the data pipeline: as mentioned in the lecture, more language pairs are available [here](https://www.manythings.org/anki/). It would be worth comparing the performance when training on separate languages, vs. combining multiple languages, and training on all of them in one go. (Language models, at scale at least, are known to benefit from the latter approach.)
3. Dig deep into the parts of the transformer:
    - Run the examples showing the mechanisms of attention, masking, and multi-head decomposition.
    - In separate cells, run the operations of `MultiHeadAttention`, `TransformerEncoder` and `TransformerDecoder` classes one by one, creating a dummy tensor of the same format as the expected input, and making sure you understand each step.
    - One interesting line of enquiry is to perform [ablation studies (scroll down a bit)](https://deeplearningwithpython.io/chapters/chapter09_convnet-architecture-patterns/#modularity-hierarchy-and-reuse): how does removing certain parts impact performance? The way to do this is to have your training and model definition pipeline ready, and change only one element: what happens if you remove the `LayerNormalization`, or place it somewhere else? What if you remove the residual connections? What if you remove or reduce the MLP after the attention part?
